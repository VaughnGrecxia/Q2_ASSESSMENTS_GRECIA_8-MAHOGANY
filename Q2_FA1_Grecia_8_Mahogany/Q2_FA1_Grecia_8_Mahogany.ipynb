```python

def addition(num1,num2):
  return num1 + num2
def subtraction(num1,num2):
  return num1 - num2
def multiplication(num1,num2):
  return num1 * num2
def division(num1,num2):
  return num1 / num2

first_number = int(input("Enter the 1st number "))
second_number = int(input("Enter the 2nd number "))

print("Choose the operation")
print("1 = addition")
print("2 = subtraction")
print("3 = multiplication")
print("4 = division")

loop = True

while loop:
  operation = int(input("Enter a number from 1-4: "))
  if operation == 1:
    print(addition(first_number, second_number))
    loop = False
  elif operation == 2:
    print(subtraction(first_number, second_number))
    loop = False
  elif operation == 3:
    print(multiplication(first_number, second_number))
    loop = False
  elif operation == 4:
    print(division(first_number, second_number))
    loop = False
  else:
    print("Invalid!")

"""Reflection
What functions did you create in your calculator program?
Addition, Subtraction, Multiplication, Division

What parameters did your functions use?
num1 and num2

What arguments were passed when the functions were called?
Adding num1 and num2 for addition, subtracting num1 and num2 for subraction, etc..

How did your program use the returned value?
passing it into print, and then displays the calculation on screen

Why is it better to divide the calculator program into functions instead of writing everything in one long block of code?
Organization and neatness of the code
"""
```
