```python
"""You are tasked with developing a number-guessing game with the following conditions:

The secret number is 17.
The player is allowed a maximum of five valid guesses.
If the player enters a negative number, the program must ignore it without deducting an attempt.
If the player guesses correctly, the program must immediately terminate the loop.
If the player uses all five valid attempts, the program displays "Game Over!"."""


Secret_number = 17
Guesses = 5

while Guesses > 0:
    Guess = int(input("Enter a number: "))

    if Guess < 0:
        print("Negative numbers are not allowed!")
        continue

    Guesses -= 1

    if Guess == Secret_number:
        print("Good job! You guessed correctly!")
        break
    else:
        print("Incorrect!")
        print("Attempts remaining:", Guesses)

if Guesses == 0 and Guess != Secret_number:
    print("Game Over!")
```
